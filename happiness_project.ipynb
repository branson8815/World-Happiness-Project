Througout this project we will be working with data collected for the World Happiness Report in 2023. This data set contains information regarding happiness and wellbeing spanning all countries from the year 2008 to the present. 



Before we begin to tackle the questions we will be facing, we should first get familiar with the data. The purpose of this data set is to determine the general happines of individuals around the world and to determine what causes the biggest impact on happiness in an individual. This data comes from a few sources: the Gallup World Poll (GWP), the World Development Indicators (WDI), the Penn World Table 10.01, and the World Health Organization's Global Health Observatory data repository. These sources are all administrative in nature, so it is safe to say that this data set is administrative data - that is, data collected by organizations for their operations which is continuously updated. This data was collected from individuals all around the world, so the population is the global population of humans. However, it is important to recognize the bias this data may present. Who had access to the poll? For poorer countries or countries with very high populations, the poll may have only have been accessed by people with access to technology or otherwise. Data could have been collected by only people who were rich enough to partake as well. We need to keep this in mind when analyzing the data. Each observation is a country's average values for each variable based upon each individual response. The variables are as follows:

### Country Name and year

These variable names are very strightforward. After briefly inspecting the data, we found that there are 165 different countries and the range of years is from 2005 to 2022.

### Life Ladder

This is the overall subjective happiness score. It is the national average response to the question of life evaluations. The reason it is 
called Life ladder is because it is a response the the question “Please imagine a ladder, with steps numbered from 0 at the
bottom to 10 at the top. The top of the ladder represents the best possible life for you and the bottom of the ladder represents the worst possible life for you.
On which step of the ladder would you say you personally feel you stand at this time?”

### Log Gdp Per Capita

The log Gdp Per capita applies the log function to the economic output of the country per person. It essentially describes the economic strength of the country and divides this among the population to estimate the individual economic well-being of citizens. As stated in the statistical appendix, "The statistics of GDP per capita (variable name gdp) in purchasing power parity (PPP) at constant 2017 international dollar prices are from World Development Indicators (WDI, version 17, metadata last updated on Jan 22, 2023). The GDP figures for Taiwan, Syria, Palestine, Venezuela, Djibouti and Yemen are from
the Penn World Table 10.01."

### Social Support

The social support variable is calculated as a proportion of people answering yes to the question of whether they feel they have social support or not. 
Specifically, this response applies to the question of "If you were in trouble, do you have relatives or friends you can count on to help you
whenever you need them, or not?”


### Healthy Life expectancy at birth

Healthy Life Expectancy at birth is fairly self-explanatory. Healthy Life expectancy, as defined by the World Health Organization (WHO) as "Average number of years that a person can expect to live in "full health" by taking into account years lived in less than full health due to disease and/or injury." The data in this table was taken from the WHO from the years of 2000, 2010, 2015, and 2019. The remaining years of healthy life expectancy are calculated via extrapolation and interpolation 

### Freedom to make life choices

This variable of Freedom to make life choices is calcuated in a similar manner to the social support variable; it is a proportion of "yes" responses. The question posed for this variable is “Are you satisfied or dissatisfied with your freedom to choose what you do with your life?”

### Generosity

Generosity in the context of this dataset is a measurement of donating to charity. Specifically it is the proportion of people answering yes to the question "have you donated to charity in the last month?" 

### Percpetions of Corruption

This variable is the average of reponses to two different questions instead of just 1 like some of the previous questions. A no corresponds to 0 and a yes corresponds to 1. The two questions are:
1)  “Is corruption widespread throughout the government or not?”
2) “Is corruption widespread within businesses or not?”

### Positive Affect

Positive Affect is basically a variable describing the positive moods or emotional states of people on average. There are three specific affect measures 
that comprise this variable: laugh, enjoyment, and doing interesting things. By answering questions about experiences of each of these affect measures in the previous day, an average can be calcuated from averaging the yes and no responses.

### Negative Affect

Negative Affect measures more negative or unpleasant emotional states. The three negative affect measures are: worry, sadnesss and anger. By answering questions about experiences of each of these affect measures in the previous day, an average can be calcuated from averaging the yes and no responses.

The following is a table which gives an organized visual representation of these variables

Name | Variable description | Type | Units of measurement
---|---|---|---
Country name | The name of the country | Nominal | None 
year | Year of collection | Numeric | None

The next question we will be answering is:

Are there any countries who have gotten at least one point happier from the years 2008 to 2022? Are there any that have gotten at least one point sadder?

To do this we must first filter the data so we only have the Life Ladder for each country in the years 2008 and 2022. Then, we must calculate the difference between the two and filter the dataset so only countries with a difference greater than 1 or less than -1 are shown.

```python
##### EXPLORATORY ANALYSIS #####
######### QUESTION TWO #########
################################

# (ii) Are there any countries who have drastically different Life Ladders between the years 2008 - 2022?
# If yes, what may have caused this? If not, why has it remained stagnant?

# Filter data for years 2008 and 2022
filtered_data = happiness[(happiness['year'] == 2008) | (happiness['year'] == 2022)]

# Group the data by country and calculate the difference
difference_data = filtered_data.pivot(index = 'Country name', columns = 'year', values = 'Life Ladder')
difference_data['difference'] = difference_data[2022] - difference_data[2008]
difference_data.reset_index(inplace = True)

# Find countries that have either increased or decreased by a point or more between the years 2008 and 2022
big_diff_data = difference_data[(difference_data['difference'] > 1) | (difference_data['difference'] < -1)]

# Rename index
big_diff_data.columns.name = None
big_diff_data.reset_index(inplace = True, drop = True)

big_diff_data
```

```python
##### EXPLORATORY ANALYSIS #####
######## QUESTION THREE ########
################################

# (iii) Does Life Ladder seem to differ between different continents? What continent has the lowest and highest Life Ladder?

# Create a function using the pycountry_convert package
def country_to_continent(country_name):
    try:
        country_alpha2 = pc.country_name_to_country_alpha2(country_name)
        country_continent_code = pc.country_alpha2_to_continent_code(country_alpha2)
        country_continent_name = pc.convert_continent_code_to_continent_name(country_continent_code)
        return country_continent_name
    except KeyError:
        return 'N/A'


# Create a new data set from happiness that includes a continent column
continent_data = happiness.copy()

continent_data['continent'] = continent_data['Country name'].apply(country_to_continent)
continent_data.head()
```

```python
# Find the mean Life Ladder for each continent
continent_grouped = continent_data.groupby(['year', 'continent']).mean(numeric_only = True)

continent_grouped = continent_grouped.reset_index()

# Remove the N/A observations
continent_grouped = continent_grouped[continent_grouped['continent'] != 'N/A']

# Remove years before 2008
continent_grouped = continent_grouped[continent_grouped['year'] >= 2008]

continent_grouped
```

## (ii) Regression Analysis

Using life ladder as our dependent variable, we can explore which variables affect the outcome the most. Linear regression is an appropriate model for this type of data considering the fact that our one dependent variable likely has a very direct relationship with the independent variables. Using multiple linear regression would be appropriate 
for a model which includes more than one independent variable. We can also fit other models which are univariate relationships between one variable and our dependent variable and explore these relationships. 

```python
#in order to have clean data for our regression model, we will drop na values

clean_happiness_data = happiness.dropna()

# our dependent variable
y = clean_happiness_data['Life Ladder']

# constructing the explanatory variable matrix using social support as our independent variable
x = sm.tools.add_constant(clean_happiness_data.loc[:, 'Social support'])



#fitting simple regression model
simple_regression = sm.OLS(endog = y, exog = x)

# store results
rslt = simple_regression.fit() 

```

```python
# print several rows of explanatory variable matrix
x.head()
```

```python
rslt.params
```

```python
rslt.cov_params()
```

```python
rslt.rsquared
```

When looking at the R squared value for our model, we see that it falls around 0.52 which means that this model is not particularly strong at explaining variation of the response variable. 

```python
clean_happiness_data.loc[:,'fitted_value'] = rslt.fittedvalues
clean_happiness_data.loc[:,'residual'] = rslt.resid
```

```python
#scatter plot to visually represent this relationhsip 

scatter_happy = alt.Chart(clean_happiness_data).mark_circle().encode(
    y = alt.Y('Life Ladder', scale = alt.Scale(zero = False), title = 'Life Ladder Score'),
    x = alt.X('Social support', scale = alt.Scale(zero = False), 
              title = 'Social support score')
    
)
scatter_happy
```

```python
#Now we are going to fit another simple regression model which looks at Life Ladder versus Log Gdp

x1 = sm.tools.add_constant(clean_happiness_data.loc[:, 'Log GDP per capita'])

simple_regression_2 = sm.OLS(endog = y, exog = x1)

rslt2 = simple_regression_2.fit()
```

```python
x1.head()
```

```python
rslt2.params
```

```python
rslt2.cov_params()
```

```python
coef_tbl = pd.DataFrame({'estimate': rslt2.params.values,
              'standard error': np.sqrt(rslt2.cov_params().values.diagonal())},
              index = x1.columns)
coef_tbl.loc['error variance', 'estimate'] = rslt2.scale

coef_tbl
```

```python
#Here we want to change our model to fit more than one independent variable
#Our variables are Healthy life expectancy at birth, Generosity, and Freedom to make life choices
x = sm.tools.add_constant(clean_happiness_data.loc[:, ['Healthy life expectancy at birth', 'Freedom to make life choices', 'Log GDP per capita']])

# fiting model and storing results
mlr = sm.OLS(endog = y, exog = x)
rslt = mlr.fit() 

#constructing a coefficient table to see how variables are related
coef_tbl_mlr = pd.DataFrame({
    'estimate': rslt.params.values,
    'standard error': np.sqrt(rslt.cov_params().values.diagonal()) 
    },
    index = x.columns 
)

coef_tbl_mlr.loc['error variance', 'estimate'] = rslt.scale

```

The above coefficients, error values, and R squared value are interesting. Additionally, the other variables in the multiple regression led to a higher R squared value of 0.7 which indicates that more of the variance of our dependent variable is being explained by including the other variables compared to log GDP alone.
